# Preprocessing


```python
import os
from PIL import Image, ImageOps
from tqdm import tqdm
```

```python

train_dir = "data/train"  
val_dir = "data/val"      


folders = ["train", "val"]  
src_base = "data"
dst_base = "transform"
target_size = 224
```

```python

# Function to count images in each class folder
def count_images_by_class(path):
    class_counts = {}
    for class_name in os.listdir(path):
        class_path = os.path.join(path, class_name)
        if os.path.isdir(class_path):
            count = len([f for f in os.listdir(class_path)
                         if f.lower().endswith((".png", ".jpg", ".jpeg", ".bmp", ".tif"))])
            class_counts[class_name] = count
    return class_counts
```

```python
def pad_and_resize(img, target_size=224, fill_color=(0,0,0)):
    w, h = img.size
    # Pad to square
    if w != h:
        if w > h:
            delta = w - h
            padding = (0, delta//2, 0, delta - delta//2)
        else:
            delta = h - w
            padding = (delta//2, 0, delta - delta//2, 0)
        img = ImageOps.expand(img, padding, fill=fill_color)
    # Resize to 224x224
    img = img.resize((target_size, target_size))
    return img
```

```python

# Count images
counts = count_images_by_class(train_dir)

# Print results
print("📊 Image counts per class in train folder:")
for class_name, num in counts.items():
    print(f"Class {class_name}: {num} images")
```

```python

# Count images
counts = count_images_by_class(val_dir)

# Print results
print("📊 Image counts per class in Validation folder:")
for class_name, num in counts.items():
    print(f"Class {class_name}: {num} images")
```

```python

for folder in folders:
    src_dir = os.path.join(src_base, folder)
    dst_dir = os.path.join(dst_base, folder)
    os.makedirs(dst_dir, exist_ok=True)

    # Process each class folder
    for class_name in os.listdir(src_dir):
        class_src = os.path.join(src_dir, class_name)
        class_dst = os.path.join(dst_dir, class_name)
        if not os.path.isdir(class_src):
            continue
        os.makedirs(class_dst, exist_ok=True)

        for file_name in tqdm(os.listdir(class_src), desc=f"Processing class {class_name}"):
            if not file_name.lower().endswith((".png", ".jpg", ".jpeg", ".bmp", ".tif")):
                continue
            src_path = os.path.join(class_src, file_name)
            dst_path = os.path.join(class_dst, file_name)
            try:
                img = Image.open(src_path)
                img_square_224 = pad_and_resize(img, target_size=224)
                img_square_224.save(dst_path)
            except Exception as e:
                print(f"⚠️ Could not process {src_path}: {e}")

print("✅ All images padded and resized to 224x224 in 'transform/train'")
```
